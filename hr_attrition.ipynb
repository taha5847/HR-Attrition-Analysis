## Feature Engineering

```python
# Create new column 'Attrition Flag' based on Attrition column

df['Attrition_Flag'] = df['Attrition'].apply(lambda x:1 if x == 'Yes' else 0)
```

```python
df['Attrition_Flag']
```

```python
df['Age'].sort_values(ascending= False).head(5)
```

```python
# Create a new column 'age group' based on age
labels = ['18-25','26-35','36-45','46-55','56-60']
bins = [17,25,35,45,55,60]
df['Age_Group'] = pd.cut(df['Age'],bins = bins, labels= labels)
```

```python
df[['Age','Age_Group']]
```

```python
df['Age_Group'].isna().sum()
```

```python
df.columns
```

```python
# Income_Band

df['MonthlyIncome'].sort_values(ascending = False)
```

```python
bins = [0, 3000, 6000, 10000, 20000]
labels = ['>3k', '3k-6k', '6k-10k', '10k-20k']

df['MonthlyIncome_Band'] = pd.cut(df['MonthlyIncome'] , bins = bins, labels = labels)
```

```python
df['MonthlyIncome_Band'].isna().sum()
```

```python
# Tenure Band

bins = [-1 ,2 ,5 ,10,20,40]
labels = ['0-2','3-5','6-10','11-20','20+']

df['Tenure_Band'] = pd.cut(df['YearsAtCompany'], bins = bins, labels = labels)
```

```python
df[['YearsAtCompany','Tenure_Band']]
```

```python
col = ['Age_Group', 'MonthlyIncome_Band','Tenure_Band']

for i in col:
    print(df[i].value_counts(),'\n\n')
    print('Null Values',df[i].isna().sum(),'\n\n')
```

```python
# Give a ratings columns to a meaningful label

rating_map = {1:'Low',2:'Medium', 3:'High', 4 : 'Very High'}
cols=['EnvironmentSatisfaction', 'JobSatisfaction','RelationshipSatisfaction', 'JobInvolvement']

for col in cols:
    df[col + '_label'] = df[col].map(rating_map)
```

```python
df[['EnvironmentSatisfaction','EnvironmentSatisfaction_label']]
```

```python
rating_map = {1: 'Bad', 2 : 'Good', 3: 'Better', 4: 'Best'}
cols = ['WorkLifeBalance']

for col in cols:
    df[col + '_label'] = df[col].map(rating_map)
```

```python
df['WorkLifeBalance'].isna().sum()
```

```python
df['PerformanceRating_label'] = df['PerformanceRating'].map({1:'Low', 2: 'Good', 3:'Excellent', 4:'Outstanding'})
```

```python
df['Education_label'] = df['Education'].map({1:'Below College', 2:'College', 3:'Bachelor', 4:'Master', 5:'Doctor'})
```
